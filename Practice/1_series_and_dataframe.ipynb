#1. What is series?
A Series is a one-dimensional labelled data structure in Pandas.
It contains a single column of data with an index.



```python
import pandas as pd

series = pd.Series([10, 20, 30, 40, 50])
print(series)
```

# 2. What is a DataFrame?

A DataFrame is a two-dimensional labelled data structure in Pandas.
It contains rows and columns, similar to a table.

```python
data = {
    "Name": ["sriyansh", "ekraj", "sanu"],
    "Age": [20, 22, 21],
    "Marks": [80, 75, 90]
}

df = pd.DataFrame(data)
print(df)
```

# 4. How can we check the shape of a Series?

```python
series = pd.Series([10, 20, 30, 40])

print(series.shape)
```

# Observation:
(4,) means the series has 4 values and is one-dimensional.

# 4. How can we check the shape of a DataFrame?


```python
data = {
    "Name": ["Subham", "Rithik", "Harsh"],
    "Marks": [80, 75, 90]
}

df = pd.DataFrame(data)

print(df.shape)
```

Observation:
(3,2) means row and 2 columns.

# 6. How can we access a column from a DataFrame as a Series?

```python
students = pd.DataFrame({
    "Name": ["Sarthak", "Ramesh", "Hrion"],
    "Marks": [80, 75, 90]
})

marks = students["Marks"]

print(marks)
print(type(marks))
```

Observation:
A single DataFrame column is returned as a Series.

# 7. How can we convert a Series into a DataFrame?

```python
marks = pd.Series([80, 75, 90])

df = marks.to_frame(name="Marks")

print(df)
```

# 8. What is the main difference between Series and DataFrame?

```python

series = pd.Series([10, 20, 30])

# DataFrame
df = pd.DataFrame({
    "A": [10, 20, 30],
    "B": [40, 50, 60]
})

print("Series:")
print(series)

print("\nDataFrame:")
print(df)
```

Observation:

Series → 1D → single column
DataFrame → 2D → multiple rows and columns
A DataFrame can contain multiple Series as columns.
