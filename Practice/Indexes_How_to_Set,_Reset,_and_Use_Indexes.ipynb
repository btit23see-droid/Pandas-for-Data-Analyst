````python
# 1️⃣ Import Pandas Library
# Import pandas to work with DataFrames
import pandas as pd
````

````python
# 2️⃣ Create a Sample DataFrame
# We will create a DataFrame with columns 'Name', 'Age', 'City'
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['Kathmandu', 'Pokhara', 'Lalitpur', 'Biratnagar']
}

df = pd.DataFrame(data)  # Create DataFrame
df
````

````python
# 3️⃣ Default Index
# By default, Pandas assigns a numeric index starting from 0
print(df.index)  # Shows the default index
````

# 4️⃣ Set a Column as Index

````python
df.set_index("Name")
````

permanently set index

````python
df.set_index("Name",inplace=True)
````

````python
df
````

````python
# 6️⃣ Reset Index

df.reset_index()

````

````python
# 6️⃣ Reset Index

df.reset_index(inplace=True)

````

### What does inplace=True mean?

- `inplace=True` means **modify the original DataFrame**
- No new DataFrame is created
- Changes happen **permanently**

````python
# 1. Create the DataFrame

import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["suman", "Siwani", "Hira", "Ganga", "jerry"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

df
````

````python
# 2. Check the default index

df.index
````

````python
# 3. Set Student_ID as the index

df.set_index("Student_ID")
````

````python
df
````

````python
# 4. Set Student_ID as the index permanently

df.set_index("Student_ID", inplace=True)

df
````

````python
# 5. Select a row using the index

df.loc[103]
````

````python
# 6. Select multiple rows using the index

df.loc[[102, 104]]
````

````python
# 7. Reset the index without inplace

df.reset_index()
````

````python
df
````

````python
# 8.Reset the index using inplace=True

df.reset_index(inplace=True)

df
````

````python
# 9. Set Name as the index

df.set_index("Name", inplace=True)

df
````

````python
# Set Student_ID as the index, display the DataFrame, then reset the index back to the default index.

df.reset_index(inplace=True)

df.set_index("Student_ID", inplace=True)

print("After setting Student_ID as index:")
print(df)

df.reset_index(inplace=True)

print("After resetting index:")
print(df)
````
